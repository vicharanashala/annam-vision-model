```python
# Importing Necessary Libraries
from ultralytics.data.split import split_classify_dataset
from ultralytics import YOLO
import albumentations as A
```

```python
# For splitting the dataset into necessary directories according to YOLO26 guidelines

data_path = "Merged_Dataset[Plant_Wild+Plant_Doc][10 Classes][For YOLO26]/train"

split_classify_dataset(data_path,train_ratio=0.8)
```

```python
# YOLO 26-L on Plant_Doc+Plant_Wild[Cropped][27 Classes]

model = YOLO("yolo26l-cls.pt") 

custom_transform = [
    A.CLAHE(clip_limit=(1, 4), tile_grid_size=(8, 8), p=1.0),
    A.HorizontalFlip(p=1.0),
    A.Rotate(limit=(0,180),p=1.0),
    A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
]

results = model.train(
    data="Merged_Dataset[Plant_Doc+Plant_Wild][All Classes]/train_split", 
    epochs=100, 
    imgsz=224, 
    batch=16,
    device=0,  # Use 0 for GPU, 'cpu' for CPU
    optimizer="AdamW",
    patience=30,
    amp=True,
    lr0=0.0001,
    augmentations=custom_transform,
    
)
```

```python
# Testing the model on Test Data

test_transform = [
    A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
]

model_loaded = YOLO("runs/classify/train9/weights/best.pt")
metrics_testing = model_loaded.val(data="Merged_Dataset[Plant_Doc+Plant_Wild][All Classes]/train_split",split='test',device=0,augmentations=test_transform)
print(metrics_testing.top1)
print(metrics_testing.top5)
```
