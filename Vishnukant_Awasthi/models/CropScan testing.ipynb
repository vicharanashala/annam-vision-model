```python
import os
import torch
import numpy as np
from PIL import Image
from tqdm import tqdm
from sklearn.metrics import classification_report, confusion_matrix

MODEL_PATH = "/kaggle/input/models/vishnu23f3003285/cropscan-rf-detr/pytorch/default/1/checkpoint_best_total.pth"

TEST_ROOT = "/kaggle/input/plantdoc-dataset/test"

CLASS_NAMES = [
    "Bell_pepper_leaf",
    "Bell_pepper_leaf_spot",
    "Potato_leaf_early_blight",
    "Potato_leaf_late_blight"
]

HEALTHY_CLASSES = ["Bell_pepper_leaf"]

CONFIDENCE_THRESHOLD = 0.27

from rfdetr import RFDETRMedium

print("Loading RF-DETR model...")
model = RFDETRMedium(pretrain_weights=MODEL_PATH)
print("Model loaded successfully!")

all_preds = []
all_labels = []

print("\nRunning evaluation...\n")

for class_name in CLASS_NAMES:

    class_folder = os.path.join(TEST_ROOT, class_name)
    if not os.path.exists(class_folder):
        continue

    for img_name in tqdm(os.listdir(class_folder)):
        img_path = os.path.join(class_folder, img_name)

        try:
            image = Image.open(img_path).convert("RGB")
        except:
            continue

        if class_name in HEALTHY_CLASSES:
            true_label = 0
        else:
            true_label = 1

        detections = model.predict(image, threshold=CONFIDENCE_THRESHOLD)

        pred_label = 1 if len(detections) > 0 else 0

        all_preds.append(pred_label)
        all_labels.append(true_label)

accuracy = np.mean(np.array(all_preds) == np.array(all_labels)) * 100

print("\n==============================")
print("CropScan Binary Test Accuracy:", round(accuracy, 2), "%")
print("==============================\n")

print("Classification Report:")
print(classification_report(all_labels, all_preds, target_names=["Healthy", "Diseased"]))

print("Confusion Matrix:")
print(confusion_matrix(all_labels, all_preds))
```
